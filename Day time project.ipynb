```python
Time_Of_Day = int(input(" Enter the number that represents the current time of day based on a 24 hour format :  "))
def Day_Time(x):
    if x >= 5 and x <= 11:
        return(" Morning ")
    elif x >= 12 and x <= 16:
        return(" Afternoon ")
    elif x >= 17 and x <= 20:
        return(" Evening ")
    elif (21 <= x <= 24) or (0 < x <= 4):
        return ("Night")
    else:
        return(" Please enter the right time. ")
print(Day_Time(Time_Of_Day))
```
